```python
#Scrap laptops'data
data={}
for p in range(1,20):
    url=f"https://webscraper.io/test-sites/e-commerce/static/computers/laptops?page={p}"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    for i in soup.find('div', class_='container test-site').find('div', class_='col-lg-9').find_all('div', class_='product-wrapper card-body'):
        title = i.find('a', class_='title').attrs['title']
        description = i.find('p', class_='description').text
        price = i.find('h4', class_='price').text[1:]
        rating = len(i.find('div', class_='ratings').find_all('span', class_='ws-icon ws-icon-star'))
        data[title] = [description,price,rating]
cols = ['title', 'description', 'price','rating']
laptop = pd.DataFrame.from_dict(data, orient='index').reset_index()
laptop.columns = cols
laptop
```

# Selenium

```python
# Create extraction function
def extract_info_selenium(url: str, country: str = 'France', season: str = '23/24') -> dict:
    # Set up Selenium options
    firefox_options = Options()
    firefox_options.add_argument("--headless")  # Run headless Firefox
    firefox_options.add_argument("--no-sandbox")
    firefox_options.add_argument("--disable-dev-shm-usage")
    firefox_options.set_preference("general.useragent.override", 
                                    "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3")
    
    # Initialize the WebDriver
    driver = webdriver.Firefox(options=firefox_options)
    
    try:
        driver.get(url)
        
        # Set selectionables
        driver.find_element(By.XPATH, "//home-away-selector//div[@class='btn-group-wrap']//label[text()='All matches']").click()
        
        # Select country
        country_dropdown = Select(driver.find_element(By.ID, 'country'))
        country_dropdown.select_by_visible_text(country)
        time.sleep(2)  # Allow time for the page to load

        # Select season
        season_dropdown = Select(driver.find_element(By.ID, 'season'))
        season_dropdown.select_by_visible_text(season)
        time.sleep(2)  # Allow time for the page to load
        
        # Fetch data
        matches = driver.find_elements(By.TAG_NAME, 'tr')
        match_list = [match.text for match in matches]
    
    finally:
        driver.quit()
    
    return match_list
```

## Custom functions

```python
from typing import Optional

def coordinate_retrieve(city_name: str, api_key: str = API_KEY) -> Optional[tuple]:
    url = "https://api.openweathermap.org/geo/1.0/direct"

    params = {"q": city_name,
              "appid": api_key,
              "lang": "en",
              'limit': 1}
    try:
        response = requests.get(url, params=params).json()
        return response[0]['lat'], response[0]['lon']
    except Exception as e:
        return (e.args[0])

```

```python
def get_current_weather(city_name, api_key: str = API_KEY) -> Optional[
    str]:
    # latitude,longitude = coordinate_retrieve(city_name, api_key)
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': city_name,
        'appid': api_key,
        'units': 'metric',  # pour obtenir les données en degrés Celsius
        'lang': 'fr'
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()
        weather = {
            'city': data['name'],
            'temperature': data['main']['temp'],
            'temp_min': data['main']['temp_min'],
            'temp_max': data['main']['temp_max'],
            'description': data['weather'][0]['description'],
            'humidity': data['main']['humidity'],
            'pressure': data['main']['pressure'],
            'wind_speed': data['wind']['speed'],
            'local_time': str(datetime.datetime.fromtimestamp(data['dt']).astimezone(
                datetime.timezone(datetime.timedelta(seconds=data['timezone']))))
        }
        return weather
    else:
        return None
```

```python
def weather_over5(city_name, api_key: str = API_KEY) -> Optional[
    pd.DataFrame]:
    url = "https://api.openweathermap.org/data/2.5/forecast"
    params = {
        'q': city_name,
        'appid': api_key,
        'units': 'metric',  # pour obtenir les données en degrés Celsius
        'lang': 'fr'
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()
        result = []
        for item in data['list']:
            weather = {
                'city': data['city']['name'],
                'temperature': item['main']['temp'],
                'temp_min': item['main']['temp_min'],
                'temp_max': item['main']['temp_max'],
                'weather': item['weather'][0]['description'],
                'humidity': item['main']['humidity'],
                'pressure': item['main']['pressure'],
                'wind_speed': item['wind']['speed'],
                'local_time': str(datetime.datetime.fromtimestamp(item['dt']).astimezone(
                    datetime.timezone(datetime.timedelta(seconds=data['city']['timezone']))))
            }
            result.append(weather)
        return pd.DataFrame.from_dict(result)
    else:
        return None
```

Quand utiliser ces outils ?

## Quand utiliser ces outils ?
- **BeautifulSoup**: 
  - Utilisé principalement pour extraire des données de fichiers HTML et XML. 
  - Idéal pour des pages statiques où le contenu ne change pas dynamiquement. 
  - Bon choix lorsque l'on a déjà le contenu HTML (par exemple, via une requête HTTP).

- **Selenium**: 
  - Utile pour interagir avec des pages web dynamiques qui utilisent JavaScript pour charger le contenu.
  - Permet d'automatiser les actions sur le navigateur (clics, remplissage de formulaires, etc.).
  - Utilisé lorsque les données sont chargées après le rendu initial de la page, comme les applications web modernes.

- **API Call**: 
  - Idéal lorsque le site web propose une API publique ou privée.
  - Permet d'accéder directement aux données sous une forme structurée (JSON, XML), souvent plus efficace et moins sujet à des restrictions légales.
  - Utilisé pour éviter le scraping en cas d'API disponible qui fournit les données nécessaires.

## Limites ?
- **BeautifulSoup**: 
  - Ne gère pas l'exécution de JavaScript, ce qui peut limiter son utilisation pour des pages web dynamiques.
  - Nécessite souvent un prétraitement pour obtenir le contenu HTML, ce qui peut nécessiter d'autres outils comme `requests`.

- **Selenium**: 
  - Plus lourd en ressources, car il nécessite un navigateur pour fonctionner. Cela peut rendre le scraping plus lent.
  - La configuration et la maintenance peuvent être plus complexes, surtout avec différents navigateurs et versions.

- **API Call**: 
  - Limité à ce que l'API expose. Certaines données peuvent ne pas être disponibles ou peuvent être soumises à des quotas d'utilisation.
  - Nécessite des connaissances sur la documentation de l'API, et les modifications de l'API peuvent casser le code existant.

## Éthiques du scraping
- **Respect des conditions d'utilisation**: Toujours lire et respecter les conditions d'utilisation des sites web. Certaines pages interdisent explicitement le scraping.
- **Impact sur le serveur**: Un scraping agressif peut surcharger le serveur du site web, affectant ainsi son fonctionnement. Il est donc essentiel de limiter le taux de requêtes (throttling).
- **Attribution des sources**: Si vous utilisez des données d'un site, il est souvent éthique de donner crédit à la source.

## Gestion des données personnelles et confidentialité
- **Respect de la vie privée**: Ne pas collecter de données personnelles sans le consentement explicite des utilisateurs. Cela inclut les informations identifiables comme les noms, adresses e-mail, etc.
- **Règlementations**: Se conformer aux lois en vigueur sur la protection des données (comme le RGPD en Europe) qui réglementent la collecte et le traitement des données personnelles.
- **Anonymisation des données**: Si des données personnelles sont collectées, envisager des méthodes d'anonymisation pour protéger l'identité des utilisateurs.
