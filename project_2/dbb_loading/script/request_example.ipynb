```python
import os
from pathlib import Path
from dotenv import load_dotenv
from pymongo import MongoClient
from sqlalchemy import create_engine, text
```

```python
try:
    load_dotenv()
    sql_usr = os.getenv("MYSQL_USER")
    sql_pwd = os.getenv("MYSQL_ROOT_PASSWORD")
    mongo_usr = os.getenv("MONGO_USR")
    mongo_pw = os.getenv("MONGO_PW")    
except Exception as e:
    exit()
```

# exemple de code requêter sur serveur mongo

```python
client= MongoClient(f'mongodb://{mongo_usr}:{mongo_pw}@localhost:27017/')
```

```python
client.list_database_names()
```

```python
db = client['sales']
```

```python
client.sales.list_collection_names()
```

```python
collection = db['clients']
```

```python
collection.find_one()
```

# Exemple de code requêter sur serveur mysql

```python
engine = create_engine(f"mysql+pymysql://{sql_usr}:{sql_pwd}@localhost/sales_details", echo=True)
```

```python
with engine.connect() as conn:
    result = conn.execute(text("select * from clients LIMIT 10"))
for row in result:
    print(row)
```
